## Label Encoding 

##### In Machine Learning, we usaully deal with the dataset with contain multiple labels in one or more columns. These labels can be in the form of the words and numbers. To make the data understandable or in human-readable form, the traning data is often labelled in words  

```python
# Machine don't understand the text  - M achine don't predict the text as well as 
# So we encode the textual data into appropriate format for the machine to learn form it and predict form it.
```

```python
# Target Variable(Attrition)
dataset["Attrition"] = dataset["Attrition"].replace({'No': 0, 'Yes': 1})
```

```python
# 1. Strip any hidden spaces first (common in CSVs)
# 2. Use .replace() instead of .map() to avoid the NaN issue if re-run
dataset['OverTime'] = dataset['OverTime'].replace({'No': 0, 'Yes': 1})
dataset['Gender'] = dataset['Gender'].replace({'Male': 0, 'Female': 1})
dataset['Over18'] = dataset['Over18'].replace({'Y': 1, 'N': 0})

```

```python
dataset.head(5)
```

```python
# encode categorical columns which are ordinal, use labelencoding
# apply labelencoder  to df_category

from sklearn.preprocessing import LabelEncoder
encoding_cols = ['BusinessTravel','Department', 'EducationField','JobRole', 'MaritalStatus' ]
label_encoder = {}
for column in encoding_cols:
    label_encoder[column] = LabelEncoder()
    dataset[column] = label_encoder[column].fit_transform(dataset[column])
```

## Machine Learning: Splitting the data into Training and Testing Sample


#### We don't use the full data for creating the model.Some data is randomly selected and kept aside for checking how good model is. This is called the Testing Data and the remaining is called the Traning Data on which the model is build. Typically 70-80 % of data used for Traning and 20-30% used for Testing Data.

```python
x = dataset.drop(["Attrition"],axis=1)  #input features
y = dataset["Attrition"].values       # output
```

## Resampling

#### In data science and statistics, resampling is a technique that involves repeatedly drawing samples from an original dataset to obtain additional information about a fitted model or a statistical estimator. 
#### It is primarily used to improve model performance, estimate the uncertainty (like confidence intervals and standard error), and validate how a model will generalize to unseen data.

#### > We are going to use Over Sampling 
#### > We are not use the Under Sampling to avoid data loss.

```python
from collections import Counter
from imblearn.over_sampling import RandomOverSampler
print(Counter(y))
y = y.astype(int)  # Or y.astype(str)
rus = RandomOverSampler(random_state= 42)
x_over, y_over = rus.fit_resample(x,y)
print(Counter(y_over))
```

```python
#Split dataset into the Traning and Testing set
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x_over, y_over,test_size = 0.2,random_state = 42)
```

```python
#Sanity check for the sampled data
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

## Logistic Regression in Machine Learning 

#### Logistic regression is a supervised machine learning algorithm used for binary classification (e.g., yes/no, 0/1) to predict the probability of a categorical outcome. It uses  
#### sigmoid function\(S(x)=\frac{1}{1+e^{-x}}\), to map linear regression outputs to a 0–1 probability range, typically classifying outcomes based on a 0.5 threshold"""

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, roc_curve, roc_auc_score
```

```python
lr = LogisticRegression()
lr.fit(x_train, y_train)
```

```python
prediction = lr.predict(x_test)
cuf_matrix = confusion_matrix(y_test, prediction)
print("Accuracy_Score - ", accuracy_score(y_test, prediction))
```

```python
(0.639676113303239)*100
```

```python


fig = plt.figure(figsize=(15, 6))

# Subplot 1: Heatmap
ax1 = fig.add_subplot(1, 2, 1)
# Pass ax1 to sns.heatmap so it knows where to draw
sns.heatmap(pd.DataFrame(cuf_matrix), cmap='Blues', annot=True, fmt='d', ax=ax1)
ax1.set_xlabel("Predicted")
ax1.set_ylabel("Expected")
ax1.set_title("Confusion Matrix")

# Subplot 2: ROC Curve
ax2 = fig.add_subplot(1, 2, 2)
prediction = lr.predict_proba(x_test)[:, 1] # Get probabilities for the positive class
fpr, tpr, _ = roc_curve(y_test, prediction) # Use probabilities, not hard predictions
auc = roc_auc_score(y_test, prediction)

# Use ax2.plot to target the second subplot
ax2.plot(fpr, tpr, label=f"Dataset, AUC={auc:.2f}")
ax2.plot([0, 1], [0, 1], 'k--') # Add diagonal reference line
ax2.set_xlabel("False Positive Rate")
ax2.set_ylabel("True Positive Rate")
ax2.set_title("ROC Curve")
ax2.legend(loc=4)

plt.tight_layout() # Prevents subplots from overlapping
plt.show()
```
